```python
import pandas as pd
import os

all_subjects = []

for filename in subject_files:
    if filename.endswith(".csv"):

        subject_id = int(filename.replace(".csv", ""))

        file_path = os.path.join(subject_folder, filename)

        subject_data = pd.read_csv(file_path)

        subject_data["Subject_ID"] = subject_id

        all_subjects.append(subject_data)

combined_data = pd.concat(all_subjects, ignore_index=True)

print("Combined dataset shape:", combined_data.shape)
print("\nColumns:")
print(combined_data.columns.tolist())

print("\nNumber of unique subjects:", combined_data["Subject_ID"].nunique())
```

```python
combined_data.head()
```

```python
from sklearn.model_selection import train_test_split

# Get all unique subject IDs
subject_ids = combined_data["Subject_ID"].unique()

# Split subjects into training and testing groups
train_subjects, test_subjects = train_test_split(
    subject_ids,
    test_size=0.20,
    random_state=42
)

print("Number of training subjects:", len(train_subjects))
print("Number of testing subjects:", len(test_subjects))
```

```python
# Create training and testing datasets using Subject_ID

train_data = combined_data[
    combined_data["Subject_ID"].isin(train_subjects)
].copy()

test_data = combined_data[
    combined_data["Subject_ID"].isin(test_subjects)
].copy()

print("Training data shape:", train_data.shape)
print("Testing data shape:", test_data.shape)

print("\nTraining subjects:", train_data["Subject_ID"].nunique())
print("Testing subjects:", test_data["Subject_ID"].nunique())
```

```python
# Select input features
X_train = train_data[
    ["Red (a.u)", "Infra Red (a.u)", "Age (year)", "Gender"]
].copy()

X_test = test_data[
    ["Red (a.u)", "Infra Red (a.u)", "Age (year)", "Gender"]
].copy()

# Select target
y_train = train_data["Hemoglobin (g/dL)"].copy()
y_test = test_data["Hemoglobin (g/dL)"].copy()

print("Training features:")
print(X_train.head())

print("\nTraining target:")
print(y_train.head())

print("\nTraining feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
```

```python
# Convert Gender into numbers

X_train["Gender"] = X_train["Gender"].map({
    "Female": 0,
    "Male": 1
})

X_test["Gender"] = X_test["Gender"].map({
    "Female": 0,
    "Male": 1
})

print("Training features after encoding:")
print(X_train.head())

print("\nGender values in training data:")
print(X_train["Gender"].value_counts())
```

```python
print("Gender values in training data:")
print(X_train["Gender"].unique())

print("\nGender values in original training data:")
print(train_data["Gender"].unique())
```

```python
# Recreate the feature datasets from the original data

X_train = train_data[
    ["Red (a.u)", "Infra Red (a.u)", "Age (year)", "Gender"]
].copy()

X_test = test_data[
    ["Red (a.u)", "Infra Red (a.u)", "Age (year)", "Gender"]
].copy()

# Encode Gender
X_train["Gender"] = X_train["Gender"].map({
    "Female": 0,
    "Male": 1
})

X_test["Gender"] = X_test["Gender"].map({
    "Female": 0,
    "Male": 1
})

print("Training features after encoding:")
print(X_train.head())

print("\nGender values:")
print(X_train["Gender"].value_counts())
```

```python
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
linear_model = LinearRegression()

# Train the model using the training data
linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
# Make predictions on the unseen testing data

y_pred = linear_model.predict(X_test)

print("First 10 actual Hemoglobin values:")
print(y_test.head(10).values)

print("\nFirst 10 predicted Hemoglobin values:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
# Create additional PPG-based features

X_train["Red_IR_Ratio"] = (
    X_train["Red (a.u)"] / X_train["Infra Red (a.u)"]
)

X_test["Red_IR_Ratio"] = (
    X_test["Red (a.u)"] / X_test["Infra Red (a.u)"]
)

X_train["Red_IR_Difference"] = (
    X_train["Red (a.u)"] - X_train["Infra Red (a.u)"]
)

X_test["Red_IR_Difference"] = (
    X_test["Red (a.u)"] - X_test["Infra Red (a.u)"]
)

X_train["Red_IR_Sum"] = (
    X_train["Red (a.u)"] + X_train["Infra Red (a.u)"]
)

X_test["Red_IR_Sum"] = (
    X_test["Red (a.u)"] + X_test["Infra Red (a.u)"]
)

print("New training feature shape:", X_train.shape)
print("\nFeatures:")
print(X_train.columns.tolist())
```

```python
from sklearn.ensemble import RandomForestRegressor

# Create Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the model
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Make predictions using Random Forest

rf_pred = rf_model.predict(X_test)

# Calculate evaluation metrics
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Results")
print("---------------------")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)
```

```python
def predict_hemoglobin(red_ppg, infrared_ppg, age, gender):

    # Encode gender
    gender_encoded = 1 if gender.lower() == "male" else 0

    # Calculate engineered features
    red_ir_ratio = red_ppg / infrared_ppg
    red_ir_difference = red_ppg - infrared_ppg
    red_ir_sum = red_ppg + infrared_ppg

    # Create input in the same order used during training
    input_data = pd.DataFrame({
        "Red (a.u)": [red_ppg],
        "Infra Red (a.u)": [infrared_ppg],
        "Age (year)": [age],
        "Gender": [gender_encoded],
        "Red_IR_Ratio": [red_ir_ratio],
        "Red_IR_Difference": [red_ir_difference],
        "Red_IR_Sum": [red_ir_sum]
    })

    # Make prediction
    prediction = rf_model.predict(input_data)

    return prediction[0]


print("Prediction function created successfully!")
```

```python
# Take one sample from the dataset for demonstration

sample = data.iloc[0]

red_value = sample["Red (a.u)"]
infrared_value = sample["Infra Red (a.u)"]
age_value = sample["Age (year)"]
gender_value = sample["Gender"]

predicted_hb = predict_hemoglobin(
    red_value,
    infrared_value,
    age_value,
    gender_value
)

print("Input Values")
print("----------------------------")
print("Red PPG       :", red_value)
print("Infrared PPG  :", infrared_value)
print("Age           :", age_value)
print("Gender        :", gender_value)

print("\nEstimated Hemoglobin")
print("----------------------------")
print(f"{predicted_hb:.2f} g/dL")

print("\nActual Hemoglobin in Dataset")
print("----------------------------")
print(f"{sample['Hemoglobin (g/dL)']:.2f} g/dL")
```

```python
import gradio as gr

def estimate_hb(red_ppg, infrared_ppg, age, gender):

    # Convert gender to the encoding used during training
    gender_encoded = 1 if gender == "Male" else 0

    # Feature engineering
    red_ir_ratio = red_ppg / infrared_ppg
    red_ir_difference = red_ppg - infrared_ppg
    red_ir_sum = red_ppg + infrared_ppg

    # Create model input
    input_data = pd.DataFrame({
        "Red (a.u)": [red_ppg],
        "Infra Red (a.u)": [infrared_ppg],
        "Age (year)": [age],
        "Gender": [gender_encoded],
        "Red_IR_Ratio": [red_ir_ratio],
        "Red_IR_Difference": [red_ir_difference],
        "Red_IR_Sum": [red_ir_sum]
    })

    # Predict hemoglobin
    prediction = rf_model.predict(input_data)[0]

    return round(prediction, 2)


print("Interface prediction function created successfully!")
```

```python
demo = gr.Interface(
    fn=estimate_hb,

    inputs=[
        gr.Number(label="Red PPG", value=115965.9),
        gr.Number(label="Infrared PPG", value=105722.9),
        gr.Number(label="Age", value=21),
        gr.Dropdown(
            choices=["Male", "Female"],
            label="Gender",
            value="Male"
        )
    ],

    outputs=gr.Number(label="Estimated Hemoglobin (g/dL)"),

    title="AI-Based Non-Invasive Hemoglobin Estimation",

    description=(
        "Enter PPG signal values, age and gender to estimate "
        "hemoglobin using the trained Random Forest model."
    )
)

print("Gradio interface created successfully!")
```
